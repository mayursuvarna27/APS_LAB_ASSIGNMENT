```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("data/rainfall_districtwise_daily_imd.csv")

# Display basic information
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

```python
# Data types and missing values
df.info()

# Statistical summary
display(df.describe())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Select relevant columns
df = df[[
    "State",
    "District",
    "Date",
    "Daily Actual",
    "Daily Normal",
    "Daily Category"
]].copy()

# Rename columns for convenience
df.rename(columns={
    "Daily Actual": "Rainfall",
    "Daily Normal": "Normal_Rainfall",
    "Daily Category": "Rainfall_Category"
}, inplace=True)

# Check for missing values
print("Missing values:\n")
print(df.isnull().sum())

# Remove rows with missing rainfall or dates
df.dropna(subset=["Date", "Rainfall"], inplace=True)

# Remove duplicate records
df.drop_duplicates(inplace=True)

print("\nShape after cleaning:", df.shape)

display(df.head())
```

```python
print("Start Date:", df["Date"].min())
print("End Date:", df["Date"].max())

print("\nNumber of States/UTs:", df["State"].nunique())
print("Number of Districts:", df["District"].nunique())

print("\nStates:")
print(sorted(df["State"].unique()))

print("\nRainfall statistics:")
display(df["Rainfall"].describe())
```

```python
state_rainfall = (
    df.groupby("State")["Rainfall"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 8))

sns.barplot(x=state_rainfall.values,
            y=state_rainfall.index)

plt.title("Average Daily Rainfall by State")
plt.xlabel("Average Rainfall (mm)")
plt.ylabel("State")

plt.show()
```

```python
# Make sure Date is in datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Extract month from Date
df["Month"] = df["Date"].dt.month

# Verify
print(df[["Date", "Month"]].head())
```

```python
# Calculate average rainfall for each state and month
monthly_state = df.pivot_table(
    values="Rainfall",
    index="State",
    columns="Month",
    aggfunc="mean"
)

plt.figure(figsize=(16, 12))

sns.heatmap(
    monthly_state,
    cmap="YlGnBu",
    linewidths=0.3,
    annot=False
)

plt.title("Monthly Rainfall Heatmap Across Indian States")
plt.xlabel("Month")
plt.ylabel("State")

plt.show()
```

What we can observe
- August and September: Higher rainfall across several states, consistent with monsoon activity.
- October and November: Rainfall generally decreases across most states.
- Mizoram, Odisha, and Sikkim: Show relatively high rainfall during some of the observed months.
- Karnataka: Rainfall decreases from August through November in this dataset.

Let's calculate a 95% confidence interval for mean daily rainfall.

```python
import scipy.stats as stats
import numpy as np

rainfall = df["Rainfall"].dropna()

n = len(rainfall)
mean_rain = rainfall.mean()
std_rain = rainfall.std()

# 95% confidence interval using t-distribution
confidence = 0.95
alpha = 1 - confidence

t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

margin_error = t_critical * (std_rain / np.sqrt(n))

lower_bound = mean_rain - margin_error
upper_bound = mean_rain + margin_error

print("Sample Size:", n)
print("Mean Rainfall:", round(mean_rain, 3), "mm")
print("95% Confidence Interval:",
      round(lower_bound, 3), "to",
      round(upper_bound, 3), "mm")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Missing-value percentage
total_values = rain.size
missing_values = int(rain.isnull().sum())

print(f"Total values: {total_values:,}")
print(f"Missing values: {missing_values:,}")
print(f"Missing percentage: {missing_values / total_values * 100:.2f}%")

# Statistics on valid rainfall observations only
valid_rain = rain.values[~np.isnan(rain.values)]

print("\nValid rainfall statistics:")
print("Mean:", np.mean(valid_rain))
print("Median:", np.median(valid_rain))
print("Standard deviation:", np.std(valid_rain))
print("95th percentile:", np.percentile(valid_rain, 95))
print("99th percentile:", np.percentile(valid_rain, 99))
```

```python
plt.figure(figsize=(10, 5))

plt.hist(valid_rain, bins=100, color="steelblue", edgecolor="white")

plt.title("Distribution of Daily Rainfall Across India (2025)")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

positive_rain = valid_rain[valid_rain > 0]

plt.hist(positive_rain, bins=100, color="teal", edgecolor="white")

plt.xscale("log")

plt.title("Distribution of Positive Daily Rainfall Across India (2025)")
plt.xlabel("Rainfall (mm) - Log Scale")
plt.ylabel("Frequency")

plt.show()
```

```python
thresholds = [0, 1, 10, 25, 50, 100]

for threshold in thresholds:
    probability = np.mean(valid_rain > threshold)

    print(
        f"P(Rainfall > {threshold} mm) = "
        f"{probability:.4f} ({probability * 100:.2f}%)"
    )
```

```python

import xarray as xr
import pandas as pd
import numpy as np
import os

# ==========================================
# 1. PATH CONFIGURATION
# ==========================================

DATA_PATH = "data/RF25_ind2025_rfp25.nc"

# ==========================================
# 2. LOAD NETCDF DATA
# ==========================================

print("Loading NetCDF rainfall dataset...")

ds = xr.open_dataset(DATA_PATH)

rain = ds["RAINFALL"]

print("\nNetCDF Dataset Information:")
print(ds)

# ==========================================
# 3. CONVERT TO PANDAS DATAFRAME
# ==========================================

print("\nConverting NetCDF to DataFrame...")

df = rain.to_dataframe().reset_index()

print("Original shape:", df.shape)

# ==========================================
# 4. DATA CLEANING
# ==========================================

print("\nCleaning dataset...")

# Remove missing rainfall observations
df = df.dropna(subset=["RAINFALL"])

# Rename rainfall column
df.rename(columns={"RAINFALL": "Rainfall"}, inplace=True)

# Remove negative rainfall values
df = df[df["Rainfall"] >= 0]

# Remove duplicate observations
df.drop_duplicates(
    subset=["TIME", "LATITUDE", "LONGITUDE"],
    inplace=True
)

# ==========================================
# 5. FEATURE ENGINEERING
# ==========================================

print("\nEngineering temporal features...")

df["Month"] = df["TIME"].dt.month

df["Day"] = df["TIME"].dt.day

df["DayOfYear"] = df["TIME"].dt.dayofyear

# Cyclical encoding of day of year
df["Day_sin"] = np.sin(
    2 * np.pi * df["DayOfYear"] / 365
)

df["Day_cos"] = np.cos(
    2 * np.pi * df["DayOfYear"] / 365
)

# ==========================================
# 6. FINAL DATASET
# ==========================================

print("\nFinal Dataset Information")
print("-" * 40)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst five rows:")
print(df.head())

print("\nStatistical Summary:")
print(df["Rainfall"].describe())

# ==========================================
# 7. SAVE PROCESSED DATA
# ==========================================

os.makedirs("data", exist_ok=True)

OUTPUT_PATH = "data/rainfall_processed.csv"

df.to_csv(OUTPUT_PATH, index=False)

print("\nProcessed dataset saved successfully!")
print("File:", OUTPUT_PATH)

# ==========================================
# 8. CLOSE NETCDF FILE
# ==========================================

ds.close()

print("\nData preparation completed!")
```
